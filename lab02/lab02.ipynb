## The model

Caffeine leaves the body at a rate proportional to how much is there, so the
amount left behind follows a **half-life** rule. If you take in $D$ milligrams
and $T$ is the half-life in hours, then after $t$ hours the amount remaining is

$$C(t) = D \cdot \left(\tfrac{1}{2}\right)^{t/T}.$$

Every $T$ hours, half of what is left disappears.

For caffeine, $T \approx 5$ hours. That is an average — it varies from person to
person, roughly 3 to 7 hours, and this one-line model ignores a great deal of
real biology. It is a model, not medical advice. It is, however, a very good
excuse to compute something.
