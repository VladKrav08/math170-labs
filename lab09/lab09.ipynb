## The model

Days are numbered through the year: January 1 is day 1, December 31 is
day 365.

On day $d$, the sun is directly overhead at the **declination**

$$\delta(d) = 23.44 \, \sin\left(\frac{2\pi\,(d + 284)}{365}\right) \quad \text{degrees}$$

— north of the equator when $\delta > 0$, south of it when $\delta < 0$.

At latitude $\varphi$ (degrees, positive north of the equator, negative south of
it), the number of hours from sunrise to sunset is

$$H = \frac{24}{\pi} \arccos\bigl(-\tan\varphi \, \tan\delta\bigr),$$

where $\tan$ needs $\varphi$ and $\delta$ in **radians**: multiply degrees by
$\pi/180$.

It is a model: it treats the sun as a point and ignores the air, which bends
sunlight around the horizon. Real days are a few minutes longer than it says.
It works for any latitude between $-66$ and $66$ degrees; closer to the poles,
on some days the sun never sets or never rises.
