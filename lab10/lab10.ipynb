## The model

Positions along the rod are measured in centimeters, from $x = 0$ at one end to
$x = 100$ at the other. We keep track of the temperature at 51 points, 2 cm
apart:

````python
# Just run this one.
x = np.linspace(0, 100, 51)
print(x)
````

An array `u` of 51 numbers is then the temperature of the whole rod at one
moment: `u[i]` is the temperature, in °C, at the point `x[i]`.

**One time step.** Heat flows from hot to cold, so each point moves toward the
temperature of its two neighbours. For every point `i` *inside* the rod,

$$u_{\text{new}}[i] = u[i] + r\,\bigl(u[i-1] - 2\,u[i] + u[i+1]\bigr),$$

where $r$ is a fixed number between 0 and $\tfrac12$ (we use $r = 0.25$). The
two **end points stay at 20 °C**: they are clamped to something at room
temperature.

It is a model: heat leaves only through the ends, not into the air. With these
numbers, for an aluminium rod one step is about one second.

## The movie (§6.4)

Nothing to write here. `animate_rod` is written for you: read it, run its
cell, then run the one after it to watch ten minutes of the rod — 600 steps of
about one second each, with $r = 0.25$. `U` has 601 rows, so the movie uses
only every 10th one, `U[::10]`.

````python
# Just run this one. It is not graded.
def animate_rod(x, U, filename):
    """
    Animate the temperature along the rod, one frame per row of U.
    Save the animation as filename (a .gif) and return it.

    x        : array of positions along the rod, in cm
    U        : 2D array, row k is the temperature after k steps
    filename : the name of the movie file, e.g. 'rod.gif'
    """
    fig = plt.figure()
    lines = plt.plot(x, U[0])            # plot the first row once
    plt.axis([x[0], x[-1], 0, 110])      # and fix the axes
    plt.xlabel('position (cm)')
    plt.ylabel('temperature (C)')

    def next_frame(u):                   # u is one row of U
        lines[0].set_ydata(u)
        return lines

    ani = FuncAnimation(fig, next_frame, frames=U, interval=50)
    ani.save(filename, writer='pillow')
    plt.close(fig)
    return ani
````
