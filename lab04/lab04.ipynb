## First, a warm-up you do not have to write

The highs above were read off a thermometer on a sunny wall, and it reads
**2 °F too high**. Before fixing anything, a careful scientist keeps the raw
numbers. Predict what the cell below prints — both lines — then run it.

````python
# Just run this one.
readings = [84.0, 88.0, 91.0, 89.0]
raw = readings                    # keep the original, just in case

for i in range(len(readings)):
    readings[i] = readings[i] - 2.0

print("corrected:", readings)
print("raw:      ", raw)
````

You kept the raw data. Python says you did not.

This is §3.3: `raw = readings` does not copy the list. It gives the **same**
list a second name, so the correction reached both. The original numbers are
gone, and nothing in the output warned you.

### Your month

Run the cell below. It puts all five functions to work on the data at the top:
fix the thermometer, then report each week.

`month_mean` wants a list of lists, so to average a single list the cell wraps
it in one more pair of brackets: `month_mean([weekend])`.

````python
# Just run this one.
offset = 2.0     # how far too high the thermometer reads, in F

ready = (to_celsius([50]) is not None and daily_swing([1], [0]) is not None
         and corrected([1.0], 1.0) is not None and month_mean([[1]]) is not None
         and split_week([1, 2, 3, 4, 5, 6, 7]) is not None)

if not ready:
    print("Finish Tasks 1-5, run their cells, then run this one again.")
else:
    fixed_month = []
    for week in month_highs:
        fixed_month.append(corrected(week, offset))

    print("         weekdays   weekend   mean swing")
    for w in range(len(fixed_month)):
        weekdays, weekend = split_week(fixed_month[w])
        swing = daily_swing(fixed_month[w], month_lows[w])
        print(f"Week {w + 1}   {month_mean([weekdays]):6.1f} F  {month_mean([weekend]):6.1f} F"
              f"   {month_mean([swing]):6.1f} F")

    print()
    mean_f = month_mean(fixed_month)
    mean_c = to_celsius([mean_f])[0]
    print(f"Average high for the month: {mean_f:.1f} F = {mean_c:.1f} C")
    print("Raw data still there?  month_highs[0] =", month_highs[0])
````
