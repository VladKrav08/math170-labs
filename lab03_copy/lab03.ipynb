## The model

Number the weeks $w = 0, 1, 2, \dots$, starting from where you are today. If
your long run today is $L_0$ miles and it grows by $p$ percent each week, then
the long run in week $w$ is

$$L(w) = L_0 \left(1 + \frac{p}{100}\right)^{w}.$$

For this lab: $L_0 = 4$ miles, $p = 10$, and the target is 13.1.

One habit from §2.5 while you type: write distances as `4.0`, not `4`. It makes
no difference in Python 3, but it is the habit that keeps you out of trouble in
every other language you will meet.

## Task 1 — Did I get there? (§2.5, §3.2)

`reached_goal(distance, goal, tol)` returns `True` when `distance` counts as
having reached `goal`, and `False` otherwise.

"Counts as" is the whole point: `distance` and `goal` are floats, so they are
allowed to disagree by a little. `tol` is how much you are willing to forgive.

One line. Note that a comparison **is already** `True` or `False` — you can hand
the comparison itself back with `return`.

````python
def reached_goal(distance, goal, tol):
    """
    True if distance has reached goal, allowing for round-off.

    distance : miles actually covered
    goal     : miles you were aiming for
    tol      : how far apart the two may be and still count as equal
    """
    return abs(goal - distance) <= tol
````

## Task 2 — How many weeks? (§3.1)

Starting from a 4-mile long run and adding 10% a week, how many weeks until the
long run is at least 13.1 miles?

`weeks_to_goal(start_miles, growth_pct, goal_miles)` returns that number of
weeks.

Before you write the loop, answer the four questions from class — as comments,
in the cell, above your code:

```
# (i)   what are the variables, and what do they start at?
# (ii)  what gets repeated?
# (iii) what condition keeps the loop going?
# (iv)  how does each variable change on one pass?
```

A week only counts once you have finished it, so a plan that starts at 4 miles
and never grows would loop forever — which is exactly what (iii) is for.

````python
def weeks_to_goal(start_miles, growth_pct, goal_miles):
    """
    Number of weeks of growth needed to reach goal_miles.

    start_miles : today's long run, in miles
    growth_pct  : percent added each week (10 means 10%)
    goal_miles  : the distance you are training for
    """
    #(i) Start_miles is the initial starting point, growth_pct is by how many percentiles the run lenght goes up every week,goal_miles is the end point of run lenght
    #(ii) growth_pct is repeated every week, since it is 10% growth every week
    #(iii) that goal_miles >= start_miles
    #(iv) it changes by 10%, changes by current_miles*(1+growth_pct)
    growth = growth_pct/100
    t = 0
    while goal_miles >= start_miles:
        start_miles += start_miles*growth
        t += 1
    


    return t

weeks_to_goal(4,10,13.1)
````

````python
# TEST -- Task 2
result = weeks_to_goal(4.0, 10, 13.1)
print(f"weeks_to_goal(4.0, 10, 13.1) = {result}")
print("expected                     = 13")

if result is None:
    print("Not implemented yet.")
elif result == 13:
    print("PASSED")
    print(f"  starting from 6 miles instead -> {weeks_to_goal(6.0, 10, 13.1)} weeks")
    print(f"  a gentler 5% a week           -> {weeks_to_goal(4.0, 5, 13.1)} weeks")
elif result == 14:
    print("One too many -- you are counting the week you started in.")
elif result == 12:
    print("One too few -- after 12 weeks the long run is still 12.55 miles.")
else:
    print("Not matching -- print distance and weeks inside the loop and watch them move.")
````

## Task 3 — The whole plan at once (§3.1, §3.3)

Task 2 threw the distances away as it went. Now keep them.

`weekly_plan(start_miles, growth_pct, weeks)` returns a **list** holding the
long run for each week, starting with week 0. For `weeks = 5` the list has five
numbers in it: weeks 0, 1, 2, 3 and 4.

Use the model from the top of the lab,

$$L(w) = L_0\left(1 + \tfrac{p}{100}\right)^{w},$$

start from an empty list, and `append` one week at a time inside a loop.

````python
def weekly_plan(start_miles, growth_pct, weeks):
    """
    The long run for each week, as a list of miles.

    start_miles : today's long run, in miles
    growth_pct  : percent added each week (10 means 10%)
    weeks       : how many weeks the list should cover
    """
    plan = []
    t = 0
    while weeks > t:
        plan.append(start_miles)
        start_miles += start_miles*(growth_pct/100)
        t+=1
    return plan
````

````python
# TEST -- Task 3
result = weekly_plan(4.0, 10, 5)
print("yours:    ", result)
print("expected: ", [4.0, 4.4, 4.84, 5.324, 5.8564], "(up to round-off)")

if result is None:
    print("Not implemented yet.")
elif not isinstance(result, list):
    print("That is not a list -- start from [] and append to it.")
elif len(result) != 5:
    print(f"Wrong length: {len(result)} weeks, expected 5. Weeks 0 through 4 is five weeks.")
elif [round(x, 4) for x in result] == [4.0, 4.4, 4.84, 5.324, 5.8564]:
    print("PASSED")
else:
    print("Right length, wrong numbers -- check the exponent. Week 0 should be")
    print("the distance you start at, unchanged.")
````

## Task 4 — How far is all of it? (§3.1, §3.3)

`total_miles(plan)` adds up every distance in the list and returns the total.

Do it with a loop that walks the list **by index**: start a counter at 0, use
`plan[i]` to reach each element, and let `len(plan)` tell you when to stop. That
is the pattern behind every sum you will write this semester.

(Python has a `sum` function. Not today — the loop is the exercise.)

````python
def total_miles(plan):
    """
    Total mileage of a plan.

    plan : a list of weekly long-run distances
    """
    d = 0
    l = len(plan)
    for i in range(l):
        d += plan[i]
    return d
````

````python
# TEST -- Task 4
result = total_miles([4.0, 4.4, 4.84])
print(f"total_miles([4.0, 4.4, 4.84]) = {result}")
print("expected                      = 13.24")

if result is None:
    print("Not implemented yet.")
elif abs(result - 13.24) < 1e-9:
    print("PASSED")
    print(f"  an empty plan -> {total_miles([])}  (should be 0)")
    print(f"  one week only -> {total_miles([7.5])}  (should be 7.5)")
else:
    print("Not matching -- does your counter start at 0, and does the loop stop")
    print("at len(plan)?")
````
